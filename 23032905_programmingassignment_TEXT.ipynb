**Project Name:** Student Finance Assistant

**Target User:** University Students Managing everyday expenses.

##


**Problem Statement**

University students often have trouble understanding how they spend their money, in managing a restricted budget and in deciding if they can afford their future expenses. Since payments are distributed among various categories, it is difficult to see where their money is being spent.

##


**Proposed Solution**

It is a finance assistant based on Python that takes transaction data in CSV format, calculates spending by category, compares this spending with a budget that has been defined by the user, and uses Gemini to answer questions regarding finance based on the results of the calculations.

##



**Success Criteria**
* It accurately calculates both the total spending and the spending by category

* Computes how much budget is left and spots any instances of overspending.

* It will not crash unexpectedly when dealing with missing or invalid data.

* Provides AI replies that are tailored and based on data.

* Offers a Gradio interface that is easy to use and can be run in Google Colab.





**Who is the user?**

The people who will use this are university students. These students need to manage their money. They have to manage their studies, their jobs and their everyday costs.

**What problem do they have?**

Students might have trouble seeing how they spend their money. They might not know how much they are spending each month. They might not know how money they have left to spend. Without knowing where the money goes small purchases can add up. This makes it harder to manage money.

**Why is this useful?**

The app helps make money information easier to understand. It takes the list of purchases. Makes a clear summary. It can help students see which areas they spend the on. It can help them track ther money and make choices about spending.

**What will the assistant do?**

The Student Finance Assistant will take money records in a CSV file. It will add up all the spending. It will sort the money into groups. It will check how much is spent against a budget the user sets. It will give explanations using Gemini. A Gradio tool will let users upload their data. They can ask questions, about how they spent their own money.


**Imputs and Outputs**

(imputs -> outputs)

Uploaded transaction CSV -> Total spending

Transaction amounts ->	Spending by category


Transaction categories ->	Remaining budget


Monthly budget	-> Budget status


User's finance question	-> Personalised AI response

##

**Potential Invalid Imputs and how the application should handle them:**

(Problem -> Handling Method)


No CSV uploaded ->	Display an error message


Empty CSV	-> Explain that there are no transactions to analyse


Missing amount column	-> Reject the file with a clear message


Non-numeric amounts	-> Exclude invalid rows and report the issue


Missing category	-> Label the transaction: Uncategorised


Negative expense amounts	-> Exclude them and report the issue


Invalid or negative budget ->	Request a valid, non-negative budget


Missing API key	-> Explain how to configure Gemini
