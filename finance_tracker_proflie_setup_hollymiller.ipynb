```python
user_name = input("What is your full name ")
theme_colour = input("What is your preferred theme colour (e.g: purple, dark) ")
primary_currency = input("What is your primary currency (e.g: AUD, USD, EUR) ")
saving_goal = input("What is your main savings goal (e.g: a holiday, car) ")

print()
print("Now let's figure out your monthly savings!")

try:
    monthly_income = float(input("What is your monthly income? "))
    monthly_necessities_spending = float(input("How much do you spend on necessities each month? "))
    goal_total_amount = float(input("How much money do you need for your saving goal? "))
    months_to_reach_goal = int(input("In how many months do you want to reach your goal? "))
except ValueError:
    print("Invalid input. Please enter numbers for income, spending, goal amount, and months. Defaulting to 0 for calculations.")
    monthly_income = 0
    monthly_necessities_spending = 0
    goal_total_amount = 0
    months_to_reach_goal = 1 # Avoid division by zero, but effectively makes saving 0

print()
print("Hello, " + user_name + ", and welcome!")
print()
print("Your Personal Finance Tracker Profile is ready. Your app theme colour will be, " + theme_colour + ", and your accounts money will be in, " + primary_currency + ".")
print()
print("We can't wait to start working towards your saving goal of, " + saving_goal + "!")

if months_to_reach_goal > 0:
    monthly_saving_required = goal_total_amount / months_to_reach_goal
    available_for_saving = monthly_income - monthly_necessities_spending

    print()
    print(f"To reach your goal of {saving_goal} ({primary_currency} {goal_total_amount:.2f}) in {months_to_reach_goal} months, you need to save {primary_currency} {monthly_saving_required:.2f} each month.")
    print(f"You currently have {primary_currency} {available_for_saving:.2f} available each month after necessities.")

    if available_for_saving >= monthly_saving_required:
        print("Great news! You have enough available to meet your monthly saving goal.")
    else:
        shortfall = monthly_saving_required - available_for_saving
        print(f"You need an additional {primary_currency} {shortfall:.2f} each month to reach your goal within the desired timeframe. Consider adjusting your spending or goal timeframe.")
else:
    print("Cannot calculate monthly savings as the number of months to reach the goal is zero or invalid.")
```
