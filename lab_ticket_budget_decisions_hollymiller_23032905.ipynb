```
Create a beginner-friendly Python program that acts as a personal purchase budget advisor. The purpose of the program is to help a user decide whether they can afford an item and, if they cannot comfortably afford it right now, explain how long it would take them to afford it.

Ask the user for:

1. Their monthly income.
2. Their monthly expenses.
3. Their current savings.
4. The price of the item they want to purchase.
5. The minimum amount of savings they want to have left after purchasing the item.

Calculate the user's disposable monthly income by subtracting their monthly expenses from their monthly income.

Then calculate how much money they would have left after buying the item using their current savings.

Use the following decision-making rules:

* If the user can afford the purchase while still keeping their chosen minimum savings amount, tell them that they can afford the purchase.
* If they can afford the purchase but it would leave them with less than their preferred savings amount, tell them that the purchase is too large for their current budget.
* If the purchase is too large for their current budget, calculate how much additional money they need to save before they can make the purchase while still keeping their chosen minimum savings amount.
* Use their disposable monthly income to calculate approximately how many months it would take them to save the additional amount needed.
* If the amount of time is not a whole number of months, round the result up to the next whole month because the user needs to have the full amount available before making the purchase.
* Tell the user the estimated number of months they need to save and explain that they could afford the purchase after this period if their income and expenses remain the same.

The program should also consider whether the purchase is a reasonable percentage of the user's monthly income:

* If the purchase is 10% or less of monthly income, describe it as a relatively small purchase.
* If the purchase is more than 10% but no more than 20% of monthly income, describe it as a moderate purchase.
* If the purchase is more than 20% of monthly income, describe it as a large purchase.

The final recommendation should be one of three outcomes:

1. “BUY IT”
   This should occur when the user can buy the item and still have at least the minimum savings amount they specified.

2. “THINK ABOUT IT”
   This should occur when the user can technically afford the item but it is a moderate or large percentage of their monthly income. Explain that they can afford it but should consider whether the purchase is worth the cost.

3. “SAVE FIRST”
   This should occur when buying the item would leave them with less than their chosen minimum savings amount or they cannot currently afford the purchase. Tell them how much more they need to save and approximately how many months it will take.

For the “SAVE FIRST” outcome, calculate the amount needed using:
amount needed = purchase price + desired savings remaining - current savings

Calculate the estimated saving time using:
months needed = amount needed / disposable monthly income

If the user has no disposable income or their monthly expenses are equal to or greater than their monthly income, explain that they currently do not have enough monthly income available to calculate a realistic saving period. Do not divide by zero.

The program should display a clear summary showing:

* Monthly income
* Monthly expenses
* Disposable monthly income
* Current savings
* Purchase price
* Desired savings remaining
* Money left after the purchase
* Purchase price as a percentage of monthly income
* Final recommendation

If the user receives “SAVE FIRST”, also display:

* How much additional money they need to save
* The estimated number of months needed to save
* A short explanation of how the calculation was made

Use clear and simple language so that someone without financial knowledge can understand the result.

Validate the inputs so that income, expenses, savings, purchase price and desired savings cannot be negative. If the user enters a negative number, display a clear error message and stop the program.

Use beginner-friendly Python code, meaningful variable names, comments where useful, and simple if/elif/else statements. Use floating-point numbers so users can enter dollar amounts with decimals.

Round all dollar amounts to two decimal places and the number of months to the nearest whole month, rounding up when necessary.

Make sure the program can be copied directly into Google Colab and run from beginning to end without errors. Do not use external libraries.

The final response should contain only the complete Python code and no explanation outside the code.
```

````python

# Paste the code your prompt produced here, then run it to show it works.

import math

def get_positive_float_input(prompt):
    """Helper function to get a positive float input from the user."""
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Error: Please enter a non-negative number.")
            else:
                return value
        except ValueError:
            print("Error: Invalid input. Please enter a number.")

print("Welcome to your Personal Purchase Budget Advisor!")
print("Please provide the following financial details:")

# Get user inputs with validation
monthly_income = get_positive_float_input("Enter your monthly income: $")
monthly_expenses = get_positive_float_input("Enter your monthly expenses: $")
current_savings = get_positive_float_input("Enter your current savings: $")
item_price = get_positive_float_input("Enter the price of the item you want to purchase: $")
desired_savings_remaining = get_positive_float_input("Enter the minimum savings you want to have left after purchase: $")

# Calculate disposable income
disposable_monthly_income = monthly_income - monthly_expenses

# Calculate savings after purchasing the item
money_left_after_purchase = current_savings - item_price

# Determine purchase percentage of monthly income
purchase_percentage = (item_price / monthly_income) * 100 if monthly_income > 0 else 0

# Initialize recommendation variables
final_recommendation = ""
additional_savings_needed = 0
estimated_months_to_save = 0

# Decision-making logic
if money_left_after_purchase >= desired_savings_remaining:
    # User can afford the purchase and keep desired savings
    if purchase_percentage <= 10:
        final_recommendation = "BUY IT"
    elif 10 < purchase_percentage <= 20:
        final_recommendation = "THINK ABOUT IT - Moderate Purchase"
    else: # purchase_percentage > 20
        final_recommendation = "THINK ABOUT IT - Large Purchase"
else:
    # Purchase is too large, need to save more or current savings are too low
    final_recommendation = "SAVE FIRST"
    # Calculate how much more is needed
    additional_savings_needed = item_price + desired_savings_remaining - current_savings

    if disposable_monthly_income <= 0:
        # Cannot calculate saving time if disposable income is zero or negative
        estimated_months_to_save = "N/A (Review your budget)"
    else:
        # Calculate estimated months to save, rounding up
        estimated_months_to_save = math.ceil(additional_savings_needed / disposable_monthly_income)

# Prepare summary output
print("\n--- Financial Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print(f"Monthly Expenses: ${monthly_expenses:.2f}")
print(f"Disposable Monthly Income: ${disposable_monthly_income:.2f}")
print(f"Current Savings: ${current_savings:.2f}")
print(f"Purchase Price: ${item_price:.2f}")
print(f"Desired Savings Remaining: ${desired_savings_remaining:.2f}")
print(f"Money Left After Purchase: ${money_left_after_purchase:.2f}")
print(f"Purchase Price as Percentage of Monthly Income: {purchase_percentage:.2f}%")

# Print final recommendation and additional details if 'SAVE FIRST'
print("\n--- Recommendation ---")
print(f"Final Recommendation: {final_recommendation}")

if final_recommendation == "SAVE FIRST":
    print(f"\nAdditional money needed to save: ${additional_savings_needed:.2f}")
    if estimated_months_to_save == "N/A (Review your budget)":
        print(f"Estimated months to save: {estimated_months_to_save}")
        print("Explanation: You currently do not have enough monthly income available to calculate a realistic saving period. Please review your budget to increase your disposable income.")
    else:
        print(f"Estimated months to save: {int(estimated_months_to_save)} months")
        print(f"Explanation: To reach your desired savings of ${desired_savings_remaining:.2f} after purchasing the item, you need to save an additional ${additional_savings_needed:.2f}. With your disposable income of ${disposable_monthly_income:.2f} per month, it would take approximately {int(estimated_months_to_save)} months.")
elif "THINK ABOUT IT" in final_recommendation:
    purchase_category = "relatively small" if purchase_percentage <= 10 else \
                        "moderate" if 10 < purchase_percentage <= 20 else "large"
    print(f"Explanation: You can technically afford this item, but it represents a {purchase_category} percentage of your monthly income. Consider whether this purchase is truly worth the cost and if it aligns with your financial goals.")
else: # BUY IT
    print(f"Explanation: Based on your income, expenses, and desired savings, you can comfortably afford this item and maintain your preferred savings. Enjoy your purchase!")

print("\nThank you for using the Personal Purchase Budget Advisor!")
````
